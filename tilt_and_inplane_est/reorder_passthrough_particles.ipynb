```python
import os
import mrcfile
import numpy as np
from pathlib import Path
from cryosparc.dataset import Dataset
```

```python
equatorial_filepath = '/Users/gerdbizi/Desktop/Rubinstein Lab/rotation_debugging_495/angle_estimated_equatorial_particles_rotation_matrix_big.cs'
old_passthrough_filepath = '/Users/gerdbizi/Desktop/Rubinstein Lab/ryan_images/J3365/J3365_passthrough_particles.cs'
output_filepath = '/Users/gerdbizi/Desktop/Rubinstein Lab/rotation_debugging_495/angle_estimated_equatorial_particles_passthrough_rotation_matrix_big.cs'
```

```python
equatorial = Dataset.load(equatorial_filepath)
passthrough = Dataset.load(old_passthrough_filepath)
```

```python
# Create a mapping from uid to index in passthrough
passthrough_uids = {uid: idx for idx, uid in enumerate(passthrough['uid'])}

reordered_indices = []
valid_equatorial_indices = []  # Track which particles to keep in equatorial dataset

# Iterate through equatorial particles and only keep those that exist in passthrough dataset
for idx, uid in enumerate(equatorial['uid']):
    if uid in passthrough_uids:
        reordered_indices.append(passthrough_uids[uid])
        valid_equatorial_indices.append(idx)
```

```python
# Create filtered datasets
filtered_equatorial = equatorial.take(valid_equatorial_indices)
passthrough_reordered = passthrough.take(reordered_indices)
```

```python
print(f"Created dataset with {len(filtered_equatorial)} particles")
print(f"Removed {len(equatorial) - len(valid_equatorial_indices)} particles that weren't in passthrough dataset")
```

```python
print(len(filtered_equatorial))
print(len(passthrough_reordered))
```

```python
filtered_equatorial.save(equatorial_filepath)
passthrough_reordered.save(output_filepath)
```

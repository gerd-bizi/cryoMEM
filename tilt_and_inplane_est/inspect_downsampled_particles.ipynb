# Inspect downsampled particles

In this notebook, we will inspect the downsampled particles to see what they look like.

```python
import mrcfile
import numpy as np
import matplotlib.pyplot as plt
import os

# First verify the file exists
file_path = '/Users/gerdbizi/Desktop/projections.128.mrc'
print(f"File exists: {os.path.exists(file_path)}")
print(f"File size: {os.path.getsize(file_path) / (1024*1024):.2f} MB")

# Load the downsampled particles with permissive mode
with mrcfile.open(file_path, permissive=True) as mrc:
    # Print file information
    print(f"Header mode: {mrc.header.mode}")
    print(f"Data shape: {mrc.data.shape if mrc.data is not None else 'No data'}")
    
    if mrc.data is None:
        print("Error: No data found in MRC file")
    else:
        # Plot the first 10 particles
        for i in range(10):
            plt.figure(figsize=(8, 8))
            plt.imshow(mrc.data[i], cmap='gray')
            plt.title(f'Particle {i+1}')
            plt.colorbar()
            plt.show()
```

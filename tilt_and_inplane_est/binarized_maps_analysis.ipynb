# Understanding Structure of Binarized Maps

In this document, we will attempt to understand the structure behind the binarized maps that were produced by the Vesicle Picker pipeline.

First, let's import the necessary libraries and load the binarized maps.

```python
import pickle as pkl
import matplotlib.pyplot as plt
import os
```

```python
binarized_maps_directory = '/Users/gerdbizi/Desktop/Rubinstein Lab/filtered_vesicles'
binarized_maps = os.listdir(binarized_maps_directory)

# Pick a random binarized map to analyze
binarized_map = binarized_maps[0]
binarized_map_path = os.path.join(binarized_maps_directory, binarized_map)
binarized_map_data = pkl.load(open(binarized_map_path, 'rb'))

# Print the keys of the binarized map data
print(binarized_map_data.keys())
```

```python
plt.imshow(binarized_map_data['composite_mask'])
plt.show()
```
